```r
#Histogram that shows distribution of regional wins % amongst entire dataset. Shows average regional wins % is around 70-75%.
options(repr.plot.width = 6, repr.plot.height = 6)
basketball_plot1 <- basketball_tidy|>
    ggplot(aes(x=reg_wins_pct))+
    geom_histogram(bins=20)+
    xlab("Regional Wins %")+
    ylab("Count of Wins %")+
    labs(title="Distribution of Regional Wins Percentage")+
    theme(text=element_text(size=14))
basketball_plot1
```

```r
#Tip for rendering plots from Canvas announcements:
options(jupyter.plot_mimetypes = c("text/plain", "image/png"))
#Create new column that shows proportions of each category in tourney_finish from 0-1.
basketball_proportions <- basketball_tidy|>
    select(tourney_finish)|>
    group_by(tourney_finish)|>
    summarize(proportion = n()/2092)

#Bar plot of each tournament finishing label vs its proportion, ordered by size.
basketball_plot2 <- basketball_proportions|>
    ggplot(aes(y= fct_reorder(tourney_finish, proportion), x=proportion))+
    geom_bar(stat="identity")+
    ylab("Tournament Finishing Place")+
    xlab("Relative Proportion")+
    labs(title="Proportion of Tournament Finishing Places")+
    theme(text=element_text(size=14))
basketball_plot2
```
